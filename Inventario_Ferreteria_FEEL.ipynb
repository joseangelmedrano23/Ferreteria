### 📌 ¿Por qué una Pila (Stack)?
Utilizamos la estructura **LIFO (Last In, First Out)**:
* **Entrada (Push):** El nuevo lote recibido se coloca en la **Cima** de la categoría (`.append()`).
* **Salida (Pop):** El producto más reciente es el primero en despacharse (`pila[-1]`). Si llega a cero stock, se elimina de la cima (`.pop()`).

```python
# ==========================================
# 1. ESTRUCTURA MULTI-PILA (DICCIONARIO DE PILAS)
# ==========================================

inventario_feel = {
    "Obra Gris": [
        {"producto": "Bolsa De Cemento Holcim 42.5kg", "cantidad": 50, "precio": 380.00},
        {"producto": "Varilla De Hierro 3/8\"", "cantidad": 100, "precio": 145.00},
        {"producto": "Bloque De Concreto 15x20x40", "cantidad": 200, "precio": 18.50}  # Cima de Obra Gris
    ],
    "Pinturas": [
        {"producto": "Pintura Aceite Blanco Galón", "cantidad": 15, "precio": 520.00},
        {"producto": "Brocha De Cerda 3\"", "cantidad": 30, "precio": 85.00},
        {"producto": "Pintura Latex Azul Cubeta", "cantidad": 8, "precio": 2100.00}    # Cima de Pinturas
    ],
    "Herramientas": [
        {"producto": "Martillo De Uña 16oz", "cantidad": 12, "precio": 220.00},
        {"producto": "Alicate Universal 8\"", "cantidad": 20, "precio": 180.00},
        {"producto": "Taladro Percutor 1/2\"", "cantidad": 5, "precio": 1650.00}     # Cima de Herramientas
    ],
    "PVC": [
        {"producto": "Tubo PVC Presión 1/2\"", "cantidad": 40, "precio": 110.00},
        {"producto": "Codo PVC 90 Grados 1/2\"", "cantidad": 80, "precio": 12.00},
        {"producto": "Pegamento PVC 4oz", "cantidad": 25, "precio": 95.00}           # Cima de PVC
    ]
}

print("✅ Base de datos cargada. Cada categoría funciona como una Pila independiente.")
```

### ⚙️ Lógica de Operaciones Principales
1. **Push (`agregar_producto`):** Inserta un elemento en la cima con validaciones de tipo y rangos estrictos (precios y cantidades $> 0$).
2. **Pop / Despacho (`despachar_producto`):** Modifica el lote superior. Si la cantidad disminuye a 0, la cima se desapila mediante `.pop()`.
3. **Peek (`mostrar_cima`):** Inspecciona únicamente el elemento en la posición `[-1]` sin alterar la estructura.

```python
# ==========================================
# 2. DEFINICIÓN DE FUNCIONES DEL SISTEMA
# ==========================================

def mostrar_menu_principal() -> None:
    print("\n" + "=" * 45)
    print("      INVENTARIO MULTI-PILA - FERRETERÍA FEEL")
    print("=" * 45)
    print("1. Registrar producto en una categoría")
    print("2. Despachar producto reciente")
    print("3. Consultar la cima de una categoría")
    print("4. Mostrar inventario completo")
    print("5. Buscar producto en almacén")
    print("6. Salir")
    print("=" * 45)


def seleccionar_categoria(categorias: dict) -> str | None:
    if not categorias:
        print("No hay categorías disponibles.")
        return None

    print("\n--- Seleccione una Categoría ---")
    llaves = list(categorias.keys())
    for idx, cat in enumerate(llaves, 1):
        print(f"{idx}. {cat}")
    print("0. Cancelar")

    while True:
        opc = input("Ingrese el número de categoría: ").strip()
        if opc == "0":
            return None
        if opc.isdigit() and 1 <= int(opc) <= len(llaves):
            return llaves[int(opc) - 1]
        print("Error: Ingrese un número de opción válido.")


def agregar_producto(categorias: dict) -> None:
    categoria = seleccionar_categoria(categorias)
    if categoria is None:
        return

    while True:
        nombre = input(f"Ingrese el nombre para [{categoria}] (o '0' para cancelar): ").strip().title()
        if nombre == "0":
            return
        if nombre and any(c.isalnum() for c in nombre):
            break
        print("Error: El nombre debe contener texto válido.")

    while True:
        try:
            cantidad = int(input("Ingrese la cantidad inicial: ").strip())
            if cantidad > 0:
                break
            print("Error: Debe ser un entero positivo mayor a 0.")
        except ValueError:
            print("Error: Entrada no válida.")

    while True:
        try:
            precio = float(input("Ingrese el precio unitario (C$): ").strip())
            if precio > 0:
                break
            print("Error: El precio debe ser mayor a 0.")
        except ValueError:
            print("Error: Entrada no válida.")

    nuevo_item = {"producto": nombre, "cantidad": cantidad, "precio": precio}
    
    # OPERACIÓN PUSH
    categorias[categoria].append(nuevo_item)
    print(f"\n¡Éxito! '{nombre}' ingresado en la CIMA de {categoria}.")


def despachar_producto(categorias: dict) -> None:
    categoria = seleccionar_categoria(categorias)
    if categoria is None:
        return

    pila = categorias[categoria]
    if not pila:
        print(f"La pila de {categoria} está vacía.")
        return

    item_cima = pila[-1]
    print(f"\nCIMA ACTUAL: {item_cima['producto']} (Disponible: {item_cima['cantidad']})")

    while True:
        entrada = input("Cantidad a despachar (0 para cancelar): ").strip()
        if entrada == "0":
            return
        try:
            a_despachar = int(entrada)
            if 0 < a_despachar <= item_cima['cantidad']:
                break
            print(f"Error: La cantidad debe estar entre 1 y {item_cima['cantidad']}.")
        except ValueError:
            print("Error: Ingrese un entero válido.")

    item_cima['cantidad'] -= a_despachar
    print(f"¡Despachadas {a_despachar} unidades de '{item_cima['producto']}'!")

    # OPERACIÓN POP (Eliminar si el stock llega a 0)
    if item_cima['cantidad'] == 0:
        pila.pop()
        print(f"El producto '{item_cima['producto']}' se ha agotado y fue removido de la CIMA.")


def mostrar_cima(categorias: dict) -> None:
    categoria = seleccionar_categoria(categorias)
    if categoria is None or not categorias[categoria]:
        print("Pila vacía o no seleccionada.")
        return

    cima = categorias[categoria][-1]
    print(f"\n--- CIMA DE [{categoria}] ---")
    print(f"Producto: {cima['producto']}")
    print(f"Stock:    {cima['cantidad']} unidades")
    print(f"Precio:   C$ {cima['precio']:.2f}")


def mostrar_inventario_completo(categorias: dict) -> None:
    print("\n" + "=" * 60)
    print("       INVENTARIO GENERAL DE FERRETERÍA FEEL")
    print("=" * 60)

    for categoria, pila in categorias.items():
        print(f"\n--- Categoría: {categoria} ({len(pila)} lotes) ---")
        if not pila:
            print("  (Pila vacía)")
        else:
            # Recorrido LIFO: Imprime desde la Cima hacia la Base
            for i, item in enumerate(reversed(pila), 1):
                subtotal = item['cantidad'] * item['precio']
                print(f"  {i}. {item['producto']} | Cant: {item['cantidad']} | Precio: C${item['precio']:.2f} | Total: C${subtotal:.2f}")


def buscar_producto(categorias: dict) -> None:
    termino = input("\nNombre del producto a buscar: ").strip().lower()
    if not termino:
        return

    encontrados = 0
    print("\n" + "=" * 60)
    print(f"          RESULTADOS PARA: '{termino.upper()}'")
    print("=" * 60)

    for categoria, pila in categorias.items():
        for pos_desde_cima, item in enumerate(reversed(pila), 1):
            if termino in item['producto'].lower():
                encontrados += 1
                subtotal = item['cantidad'] * item['precio']
                print(f"► Categoría: [{categoria}] | Posición desde Cima: {pos_desde_cima}")
                print(f"  Producto: {item['producto']} | Stock: {item['cantidad']} | Total: C$ {subtotal:.2f}")
                print("-" * 60)

    if encontrados == 0:
        print(f"No se encontraron coincidencias para '{termino}'.")
```

### 🚀 Demostración Interactiva
Ejecuta la siguiente celda para iniciar el menú principal interactivamente durante la exposición.

```python
# Bucle interactivo para mostrar la aplicación en vivo
while True:
    mostrar_menu_principal()
    opcion = input("Seleccione una opción (1-6): ").strip()

    if opcion == "1":
        agregar_producto(inventario_feel)
    elif opcion == "2":
        despachar_producto(inventario_feel)
    elif opcion == "3":
        mostrar_cima(inventario_feel)
    elif opcion == "4":
        mostrar_inventario_completo(inventario_feel)
    elif opcion == "5":
        buscar_producto(inventario_feel)
    elif opcion == "6":
        print("Saliendo del sistema de inventario FEEL...")
        break
    else:
        print("Opción no válida. Seleccione un número del 1 al 6.")
```
